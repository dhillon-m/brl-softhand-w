```python
##############################
# Import necessary libraries #
##############################

import serial
import serial.tools.list_ports
import time
import numpy as np

import soft_wrist_lib as swl
```

```python
################################
# Setup wrist and hand control #
################################

# List available serial ports
ports = serial.tools.list_ports.comports()
for port in ports:
    print(port.device)

# Open serial port
ser = serial.Serial('COM7', baudrate=1000000, timeout=0.1)


###############
# Test Script #
###############

swl.home_hand(ser)

swl.move_wrist(ser, 0, 60)
swl.move_wrist(ser, 0, -60)
swl.move_wrist(ser, 90, 0)
swl.move_wrist(ser, -90, 0)

swl.home_hand(ser)
    
# Close the serial port
ser.close()
```
